# RAAI-SGC 2.0 — Dataset EDA (run locally)


```python
from pathlib import Path
import csv
from collections import Counter
master = Path('data/processed/master.csv') if Path('data/processed/master.csv').exists() else Path('../data/processed/master.csv')
with master.open(encoding='utf-8-sig', newline='') as f:
    rows = list(csv.DictReader(f))
print('Record types:', Counter(r['record_type'] for r in rows))
print('Domains (sources):', Counter(r['domain'] for r in rows if r['record_type']=='SOURCE'))
print('Split (sources):', Counter(r['split'] for r in rows if r['record_type']=='SOURCE'))
print('Unreviewed Q&A:', sum(r['record_type']=='QA_DRAFT' and r['review_status']!='APPROVED' for r in rows))
```

```python
# Check evidence linkage and report statistics; no inferred factual approvals.
source_ids = {r['source_id'] for r in rows if r['record_type']=='SOURCE'}
print('Orphan rows:',sum(r['source_id'] not in source_ids for r in rows))
print('Unique graph terms:', len({r['target_concept'] for r in rows if r['record_type']=='EDGE'}))
print('Missing source URLs:',sum(r['record_type']=='SOURCE' and not r['source_url'] for r in rows))
```
